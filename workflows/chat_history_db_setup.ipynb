```python
import psycopg2
from helpers.db_utils import get_history_chat_connection_uri

db_chat_history_uri = get_history_chat_connection_uri()
chat_history_conn = psycopg2.connect(db_chat_history_uri)
cur = chat_history_conn.cursor()

cur.execute("""
  CREATE TABLE user_chats (
  session_id VARCHAR PRIMARY KEY,
  customer_id INT,
  messages JSONB
);
""")

chat_history_conn.commit()

cur.close()
chat_history_conn.close()
```
