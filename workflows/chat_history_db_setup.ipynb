# 🚀 Chat History Database Setup Notebook

Welcome! This notebook will guide you through setting up the **chat history database** for storing user chat sessions.  
You'll find code to connect to your PostgreSQL database, create the necessary tables, and ensure your environment is ready for chat data storage.

**Instructions:**
- Make sure your environment variables are configured as described in `docs/env_file_configuration.md`.

Let's get started! 💬🗄️

```python
# This notebook sets up the chat history database table for storing user chat sessions
# It connects to the PostgreSQL database and creates the user_chats table if it does not exist

import psycopg2
from helpers.db_utils import get_history_chat_connection_uri

# Get the connection URI for the chat history database
# Make sure your environment variables are set up as described in docs/env_file_configuration.md

db_chat_history_uri = get_history_chat_connection_uri()
chat_history_conn = psycopg2.connect(db_chat_history_uri)
cur = chat_history_conn.cursor()

# Create the user_chats table to store chat sessions, customer IDs, and messages as JSONB
cur.execute("""
  CREATE TABLE IF NOT EXISTS user_chats (
    session_id VARCHAR PRIMARY KEY,  -- Unique session identifier
    customer_id INT,                 -- Customer/user ID
    messages JSONB                   -- Chat messages stored as JSONB
  );
""")

chat_history_conn.commit()

cur.close()
chat_history_conn.close()
```
